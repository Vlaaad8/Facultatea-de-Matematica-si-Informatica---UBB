```python
from random import sample
from math import perm, comb
from itertools import permutations, combinations

def aranjamente(cuvant,number,numar_total=False,aleator=False):
    if aleator==True:
        return sample(cuvant,number)
    elif numar_total==True:
        return len(list(permutations(cuvant,number)))
    else:
        return list(permutations(cuvant,number))

def combinari(cuvant,number,numar_total=False,aleator=False):
    if aleator==True:
        return sample(list(combinations(cuvant,number)),1)
    elif numar_total==True:
        return int(len(list(permutations(cuvant,number)))/factorial(number))
    else:
        return list(permutations(cuvant,number))
    
print("Rezultate aranjamente:")
print("[1]")
print(aranjamente('word',2))
print("[2]")
print(aranjamente('word',2,numar_total=True))
print("[3]")
print(aranjamente('word',2,aleator=True))
print("Rezultate combinari:")
print("[4]")
print(combinari('word',2))
print("[5]")
print(combinari('word',2,numar_total=True))
print("[6]")
print(combinari('word',2,aleator=True))
```
