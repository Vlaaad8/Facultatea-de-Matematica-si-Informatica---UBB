```python
from random import randint


def sameBirthday(nr_pers, simulations):
    fav = 0
    for i in range(simulations):
        persoane = []
        for i in range(nr_pers):
            persoane.append(randint(1, 365))
        if len(set(persoane)) != nr_pers:
            fav += 1
    return fav / simulations


n = int(input("Numar persoane:"))
simulari = int(input("Numar de simulari:"))
print(sameBirthday(n, simulari));
```

```python
def sameBirthday2(nrPersoane):
    p = 1;
    for i in range(nrPersoane):
        p = ((365 - i) / 365) * p;
    return 1 - p


n = int(input("Numar persoane:"))
print(sameBirthday2(n))
```

```python
from matplotlib.pyplot import plot, grid, title, show
from matplotlib.pyplot import axis, plot, show
def sameBirthday(nr_pers, simulations):
    fav = 0
    for i in range(simulations):
        persoane = []
        for i in range(nr_pers):
            persoane.append(randint(1, 365))
        if len(set(persoane)) != nr_pers:
            fav += 1
    return fav / simulations

n=int(input("Numar persoane:"))
x = range(1,50)
title("Plot")
y=[sameBirthday(n,10000) for n in x]
y1=[sameBirthday2(n) for n in x]
plot(x,y,'g*')
plot(x,y1,'r*')
grid()
show()
```
